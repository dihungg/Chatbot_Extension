# **Notebook template để preprocess dữ liệu dạng bảng cho các project data science**

## **Import các thư viện cần thiết**

## **Chuẩn bị dữ liệu thô - Data preparation**
- Đầu vào: Nguồn đến dữ liệu (API, database, v.v.)
- Đầu ra: Dữ liệu thô trên máy local hoặc là function, class Python để load dữ liệu tự động từ nguồn (streaming hoặc batching)

## **Làm sạch từng bảng dữ liệu thô - Data cleaning**
- Đầu vào: Các bảng dữ liệu thô
- Đầu ra: Các bảng dữ liệu đã được làm sạch
    - Tiêu chí sạch:
        - [ ] No missing data: Không có mẫu nào bị thiếu dữ liệu - giá trị NaN, null, v.v. ở các biến
            - [ ] Tính tỉ lệ missing của mỗi cột trong từng bảng
            - [ ] Dựa vào ngữ cảnh từng bảng để đoán xem vì sao dữ liệu bị thiếu
            - [ ] Dùng các phương pháp kiểm định thống kê để kiểm tra cơ chế mất dữ liệu của mỗi cột là MCAR, MAR hay MNAR, v.v. Kết hợp với thông tin ngữ cảnh, kiến thức nghiệp vụ để hình thành giả định về nguyên nhân mất dữ liệu
            - [ ] ...
        - [ ] No row-duplication within each table: Không có các mẫu dữ liệu trùng lắp trong nội bộ mỗi bảng
        - [ ] Standardized formats: Format ngày tháng năm, chữ viết thuờng viết hoa, format của các cột dữ liệu liên quan đến domain. v.v. phải được chuẩn hoá (các dòng trong cùng cột phải giống nhau và phải tuân theo quy chuẩn)
        - [ ] No logical inconsistencies:
            - [ ] Logic về kiểu dữ liệu trong 1 cột: Tên phải là str, tuổi phải là int hoặc float, số điện thoại phải là str, địa chỉ phải là str, v.v.
            - [ ] Logic về giá trị của dữ liệu trong một cột: Tên thì xử lý linh hoạt theo tình huống, tuổi thì phải là số dương bé hơn tuổi giới hạn sinh học, số điện thoại thì chỉ chứa chữ số, v.v.
            - [ ] Logic về giá trị của dữ liệu giữa các cột: Xét trong cùng 1 dòng, nếu cột A có giá trị này thì cột B phải có giá trị nào, không được có giá trị nào?
            - [ ] Các logic nghiệp vụ khác
            - [ ] ...

## **Biến đổi dữ liệu thô thành format phù hợp để phân tích, mô hình hoá**
- Đầu vào: Tập dữ liệu gồm các bảng dữ liệu thống nhất giữa các nguồn dữ liệu khác nhau
- Đầu ra: Dữ liệu ở định dạng mà mô hình có thể tính toán được và tính toán hiệu quả
    - [ ] Encode các cột dữ liệu text, categorical thành dạng số học (numerical)
    - [ ] Các cột dữ liệu gốc ở dạng numerical thì phải được chuẩn hoá bằng các phương pháp như: scale về 1 miền giá trị, normalize dữ liệu (giả sử dữ liệu gần với phân phối chuẩn), v.v.
    - [ ] Quyết định bộ dữ liệu cuối cùng sẽ có các feature gì: Các variable nào được giữ nguyên làm 1 feature? Các variable nào có thể được kết hợp lại thành 1 feature? Các variable nào có thể không cần dùng làm feature?
    - [ ] ...
